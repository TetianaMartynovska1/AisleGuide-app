```r
library(ggplot2)
herbivores <- msleep[msleep$vore == "herbi",]
n1 <- sum(!is.na(herbivores$sleep_total))
x1 <- sum(herbivores$sleep_total > 7, na.rm = TRUE)

carnivores <- msleep[msleep$vore == "carni",]
n2 <- sum(!is.na(carnivores$sleep_total))
x2 <- sum(carnivores$sleep_total > 7, na.rm = TRUE)
```

```r
library(dplyr)

msleep |>
    filter(vore == "herbi") |>
    select(sleep_total) |>
    na.omit() |>
    nrow() -> n1

msleep |>
    filter(vore == "herbi",
           sleep_total > 7) |>
    count() |>
    pull() -> x1

msleep |>
    filter(vore == "carni") |>
    select(sleep_total) |>
    na.omit() |>
    nrow() -> n2

msleep |>
    filter(vore == "carni",
           sleep_total > 7) |>
    count() |>
    pull() -> x2
```

```r
p1 <- x1 / n1
p2 <- x2 / n2

diff <- p1 - p2

se <- sqrt(p1*(1-p1)/n1 + p2*(1-p2)/n2)

ci_low <- diff - 1.96 * se
ci_high <- diff + 1.96 * se

c(diff, ci_low, ci_high)
```
